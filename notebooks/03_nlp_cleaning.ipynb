# Nettoyage NLP - Prétraitement des Textes

Notebook simple pour débutants : nettoyage et préparation des textes

## 1. Import des bibliothèques

```python
import pandas as pd
import re
import nltk
from nltk.corpus import stopwords
from nltk.tokenize import word_tokenize

# Télécharger les données nécessaires pour NLTK (à exécuter une fois)
nltk.download('punkt')
nltk.download('punkt_tab')
nltk.download('stopwords')

# Charger les données
df = pd.read_csv('../data/processed/clean1.csv')

# Afficher un exemple de texte
print("Exemple de texte brut:")
print(df['body'].iloc[0][:200])
```

```python
df.head()
```

## 2. Normalisation (minuscules)

```python
# Convertir tout en minuscules
# Exemple: "Hello World" devient "hello world"

texte_exemple = "Hello World! This Is An EXAMPLE."
texte_minuscule = texte_exemple.lower()

print("Avant normalisation:")
print(texte_exemple)
print()
print("Après normalisation (minuscules):")
print(texte_minuscule)
```

```python
# Appliquer sur nos données
df['text_clean'] = df['body'].fillna('').str.lower()

print("Exemple après normalisation:")
print(df['text_clean'].iloc[0][:200])
```

## 3. Suppression de la ponctuation

```python
# Supprimer la ponctuation
# On garde seulement les lettres et les chiffres

texte_exemple = "Hello, world! How are you? I'm fine, thanks..."

# Remplacer tout ce qui n'est pas lettre ou chiffre par un espace
texte_sans_ponctuation = re.sub(r'[^a-zA-Z0-9\s]', ' ', texte_exemple)

print("Avant suppression:")
print(texte_exemple)
print()
print("Après suppression de ponctuation:")
print(texte_sans_ponctuation)
```

```python
# Fonction pour supprimer la ponctuation
def remove_punctuation(text):
    """Supprimer la ponctuation d'un texte"""
    return re.sub(r'[^a-zA-Z0-9\s]', ' ', text)

# Appliquer sur nos données
df['text_clean'] = df['text_clean'].apply(remove_punctuation)

print("Exemple après suppression de ponctuation:")
print(df['text_clean'].iloc[0][:200])
```

## 4. Tokenisation

```python
# Tokenisation = découper le texte en mots
# Exemple: "hello world" devient ["hello", "world"]

texte_exemple = "hello world how are you"
tokens = word_tokenize(texte_exemple)

print("Texte original:")
print(texte_exemple)
print()
print("Tokens (mots):")
print(tokens)
print(f"Nombre de mots: {len(tokens)}")
```

```python
# Appliquer la tokenisation sur nos données
df['tokens'] = df['text_clean'].apply(word_tokenize)

print("Exemple de tokens:")
print(df['tokens'].iloc[0][:20])  # Afficher les 20 premiers mots
print(f"\nNombre total de mots dans ce texte: {len(df['tokens'].iloc[0])}")
```

## 5. Suppression des stopwords

```python
# Stopwords = mots vides (très communs, peu informatifs)
# Exemples: the, is, at, which, on, a, an...

# Charger les stopwords anglais
stop_words = set(stopwords.words('english'))

print("Exemples de stopwords (anglais):")
print(list(stop_words)[:20])  # Afficher 20 stopwords
```

```python
# Exemple de suppression des stopwords
texte_exemple = "the quick brown fox jumps over the lazy dog"
tokens = word_tokenize(texte_exemple)

# Garder seulement les mots qui ne sont pas des stopwords
tokens_sans_stopwords = [word for word in tokens if word not in stop_words]

print("Avant suppression des stopwords:")
print(tokens)
print()
print("Après suppression des stopwords:")
print(tokens_sans_stopwords)
```

```python
# Fonction pour supprimer les stopwords
def remove_stopwords(tokens):
    """Supprimer les stopwords d'une liste de tokens"""
    return [word for word in tokens if word not in stop_words and len(word) > 2]

# Appliquer sur nos données
df['tokens_clean'] = df['tokens'].apply(remove_stopwords)

print("Exemple avant suppression des stopwords:")
print(df['tokens'].iloc[0][:15])
print()
print("Exemple après suppression des stopwords:")
print(df['tokens_clean'].iloc[0][:15])
print()
print(f"Nombre de mots avant: {len(df['tokens'].iloc[0])}")
print(f"Nombre de mots après: {len(df['tokens_clean'].iloc[0])}")
```

## 6. Pipeline complète de nettoyage

```python
# Fonction complète qui fait tout le nettoyage
def clean_text(text):
    """
    Nettoyer un texte complet:
    1. Minuscules
    2. Suppression ponctuation
    3. Tokenisation
    4. Suppression stopwords
    """
    # Étape 1: Minuscules
    text = text.lower()
    
    # Étape 2: Suppression ponctuation
    text = re.sub(r'[^a-zA-Z0-9\s]', ' ', text)
    
    # Étape 3: Tokenisation
    tokens = word_tokenize(text)
    
    # Étape 4: Suppression stopwords et mots courts
    tokens = [word for word in tokens if word not in stop_words and len(word) > 2]
    
    return tokens

# Test sur un exemple
texte_test = df['body'].iloc[0]
print("Texte original:")
print(texte_test[:150])
print("\n" + "="*70 + "\n")

texte_nettoye = clean_text(texte_test)
print("Texte nettoyé (tokens):")
print(texte_nettoye[:20])
```

```python
# Appliquer sur tous les textes
print("Application du nettoyage sur tous les tickets...")
df['text_cleaned'] = df['body'].fillna('').apply(clean_text)

# Convertir les tokens en texte (pour certains modèles)
df['text_final'] = df['text_cleaned'].apply(lambda x: ' '.join(x))

print("\nAperçu des résultats:")
print("="*70)
for i in range(3):
    print(f"\nTicket {i+1}:")
    print(f"Original: {df['body'].iloc[i][:100]}...")
    print(f"Nettoyé:  {df['text_final'].iloc[i][:100]}...")
```

## 7. Sauvegarde des données nettoyées

```python
# Sauvegarder le DataFrame avec les textes nettoyés
# Décommenter pour sauvegarder

df[['type', 'text_final', 'tokens_clean']].to_csv('../data/processed/clean2.csv', index=False)
print("Données sauvegardées dans: ../data/processed/clean2.csv")

print("Colonnes disponibles:")
print(df[['type', 'text_final']].head())
```
